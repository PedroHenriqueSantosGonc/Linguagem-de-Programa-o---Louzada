```python
base = pd.read_csv('../brasileirao.csv')
base.head()
```

Passo 2: Calcular o saldo de gols(GC)

```python
base['Saldo de Gols'] = base['GP'] - base['GC']
base.head()
```

```python
base['Pontos'] = (base['V'] * 3) + (base['E'] * 1)
base.head(20)
```

```python
clube_40_pontos = base[base['Pontos'] > 40]

clube_40_pontos
```

```python
base.plot(x='Clube', y='Pontos', kind='bar', title='Pontuação dos Clubes no Brasileirão 2024')
```

```python
melhor_saldo =  base[base['Saldo de Gols'] == base['Saldo de Gols'].max()]

melhor_saldo
```

```python
top_vitorias = base.sort_values(by='V', ascending=False).head(5)

top_vitorias
```
